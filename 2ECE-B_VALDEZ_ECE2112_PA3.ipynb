### **Problem B. Model Lookup**

### **Problem C. Multi-Model Subsetting**

```python
selected_cars = cars.loc[(cars['Model'] == 'Datsun 710') |
                         (cars['Model'] == 'Lotus Europa') |
                         (cars['Model'] == 'Ferrari Dino'),
                         ['Model', 'mpg', 'cyl', 'hp', 'gear']]
selected_cars
```
